```python
placement_counts = df["Placement"].value_counts()

placement_counts.plot(kind="bar")

plt.title("Student Placement Distribution")
plt.xlabel("Placement Status")
plt.ylabel("Number of Students")
plt.show()
```

```python
placement_counts.plot(kind="pie", autopct="%1.1f%%")

plt.title("Placement Percentage Distribution")
plt.ylabel("")
plt.show()
```

```python
cgpa_by_placement = df.groupby("Placement")["CGPA"].mean()

cgpa_by_placement.plot(kind="bar")

plt.title("Average CGPA by Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("Average CGPA")
plt.show()
```

```python
internship_placement = pd.crosstab(
    df["Internship_Experience"],
    df["Placement"],
    normalize="index"
) * 100

internship_placement["Yes"].plot(kind="bar")

plt.title("Placement Rate by Internship Experience")
plt.xlabel("Internship Experience")
plt.ylabel("Placement Rate (%)")
plt.show()
```

```python
project_placement = pd.crosstab(
    df["Projects_Completed"],
    df["Placement"],
    normalize="index"
) * 100

project_placement["Yes"].plot(kind="bar")

plt.title("Placement Rate by Number of Projects")
plt.xlabel("Projects Completed")
plt.ylabel("Placement Rate (%)")
plt.show()
```

```python
communication_by_placement = df.groupby("Placement")["Communication_Skills"].mean()

communication_by_placement.plot(kind="bar")

plt.title("Average Communication Skills by Placement")
plt.xlabel("Placement Status")
plt.ylabel("Average Communication Skills Score")
plt.show()
```
